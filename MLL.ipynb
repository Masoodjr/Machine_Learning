```python
import pandas as pd
import os

def create_csv_files(input_file, output_dir='output_csv'):
    # Create output directory if it doesn't exist
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    try:
        # Read the main database file
        df = pd.read_csv(input_file)
        
        # Clean up column names (remove extra spaces)
        df.columns = [col.strip() for col in df.columns]
        
        # Create Universities CSV
        universities = df[['University ID', 'University Name', 'Location', 'Application Deadline', 'Ranking', 'Placement Rate (%)']].copy()
        universities = universities.drop_duplicates(subset=['University ID'])
        universities.rename(columns={
            'University ID': 'university_id',
            'University Name': 'university_name',
            'Location': 'location',
            'Application Deadline': 'application_deadline',
            'Ranking': 'ranking',
            'Placement Rate (%)': 'placement_rate'
        }, inplace=True)
        universities.to_csv(f'{output_dir}/universities.csv', index=False)
        print(f"Created universities.csv with {len(universities)} records")
        
        # Create Students CSV
        students = df[['Student First Name', 'Student Last Name', 'Student Email', 'GRE Score', 'TOEFL Score', 'GPA', 'Phone', 'Date of Birth']].copy()
        students = students.drop_duplicates(subset=['Student Email'])
        students['student_id'] = range(1, len(students) + 1)
        students.rename(columns={
            'Student First Name': 'first_name',
            'Student Last Name': 'last_name',
            'Student Email': 'email',
            'GRE Score': 'gre_score',
            'TOEFL Score': 'toefl_score',
            'GPA': 'gpa',
            'Phone': 'phone',
            'Date of Birth': 'date_of_birth'
        }, inplace=True)
        cols = ['student_id', 'first_name', 'last_name', 'email', 'gre_score', 'toefl_score', 'gpa', 'phone', 'date_of_birth']
        students = students[cols]
        students.to_csv(f'{output_dir}/students.csv', index=False)
        print(f"Created students.csv with {len(students)} records")
        
        # Create Programs CSV
        programs = df[['University ID', 'Program Name', 'Program Type', 'Program Duration (Years)', 'Tuition Fee (USD/year)', 'Average Pay (USD/year)']].copy()
        programs = programs.drop_duplicates(subset=['University ID', 'Program Name'])
        programs['program_id'] = range(1, len(programs) + 1)
        programs.rename(columns={
            'University ID': 'university_id',
            'Program Name': 'program_name',
            'Program Type': 'program_type',
            'Program Duration (Years)': 'duration',
            'Tuition Fee (USD/year)': 'tuition_fee',
            'Average Pay (USD/year)': 'average_pay'
        }, inplace=True)
        cols = ['program_id', 'university_id', 'program_name', 'program_type', 'duration', 'tuition_fee', 'average_pay']
        programs = programs[cols]
        programs.to_csv(f'{output_dir}/programs.csv', index=False)
        print(f"Created programs.csv with {len(programs)} records")
        
        # Create Funding CSV
        funding = df[['University ID', 'Funding Source', 'Funding Amount', 'Funding Year']].copy()
        funding = funding.drop_duplicates()
        funding['funding_id'] = range(1, len(funding) + 1)
        funding.rename(columns={
            'University ID': 'university_id',
            'Funding Source': 'funding_source',
            'Funding Amount': 'amount',
            'Funding Year': 'funding_year'
        }, inplace=True)
        cols = ['funding_id', 'university_id', 'funding_source', 'amount', 'funding_year']
        funding = funding[cols]
        funding.to_csv(f'{output_dir}/funding.csv', index=False)
        print(f"Created funding.csv with {len(funding)} records")
        
        # Create Users CSV - dummy data based on students
        # Assuming each student has a user account with role=1
        users = pd.DataFrame({
            'user_id': range(1, len(students) + 1),
            'student_id': students['student_id'].values,
            'role': 1
        })
        users.to_csv(f'{output_dir}/users.csv', index=False)
        print(f"Created users.csv with {len(users)} records")
        
        # Create Wishlist CSV - create dummy wishlist data
        # For demonstration, we'll create wishlist entries by pairing students with programs
        wishlist_data = []
        for student_id in range(1, min(len(students) + 1, 10)):  # Limit to avoid too many entries
            for program_id in range(1, min(4, len(programs) + 1)):  # Each student gets up to 3 wishlist items
                wishlist_data.append({
                    'wishlist_id': len(wishlist_data) + 1,
                    'student_id': student_id,
                    'program_id': program_id
                })
        
        wishlist = pd.DataFrame(wishlist_data)
        wishlist.to_csv(f'{output_dir}/wishlist.csv', index=False)
        print(f"Created wishlist.csv with {len(wishlist)} records")
        
        print(f"All CSV files have been created in the '{output_dir}' directory")
        
    except Exception as e:
        print(f"Error processing the database file: {e}")

if _name_ == "_main_":
    # Replace with your actual input file path
    create_csv_files("main_database.csv")
```
